```python
import numpy as np

# Task 1: 5x5 matrix with border 1, interior 0
matrix = np.zeros
((5, 5))
matrix[0, :] = 1
matrix[-1, :] = 1
matrix[:, 0] = 1
matrix[:, -1] = 1
print("Task 1 - 5x5 Matrix:")
print(matrix)
print()

# Task 2: Normalize random array
np.random.seed(42)
random_data = np.random.randn(100, 3)
normalized_data = (random_data - random_data.mean(axis=0)) / random_data.std(axis=0)
print("Task 2 - Normalization:")
print(f"Original mean: {random_data[:, 0].mean():.4f}, std: {random_data[:, 0].std():.4f}")
print(f"Normalized mean: {normalized_data[:, 0].mean():.4f}, std: {normalized_data[:, 0].std():.4f}")
print()

# Task 3: Linear regression with normal equation
np.random.seed(42)
X = np.random.randn(50, 3)
true_theta = np.array([2.5, -1.2, 3.7])
y = X @ true_theta + np.random.randn(50) * 0.1
theta_hat = np.linalg.inv(X.T @ X) @ X.T @ y
print("Task 3 - Linear Regression:")
print(f"True theta: {true_theta}")
print(f"Estimated theta: {theta_hat.round(4)}")
```
